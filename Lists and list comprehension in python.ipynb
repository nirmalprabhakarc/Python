list comprehension syntax

basic syntax [expression for item in iteration]

with conditional logic [expression for item in condition]

```python
#basic list comprehension
square= [x**2 for x in range(10)]
print(square)
```

```python
#specific condition
even_square = [x**2 for x in range(10) if x%2==0]
print(even_square)
```

```python
#nested list comprehension
lst1 =[1,2,3,4]
lst2 = ["a","b","c","d"]
pair = [[i,j]for i in lst1 for j in lst2]
print(pair)
```

```python
#list comprehension with function calls
words = ["hello","world","python","list","compression"]
length = [len(word) for word in words]
print(length)
```
