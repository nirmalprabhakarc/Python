```python
#while loop

count = 0

while count<5:
  print(count)
  count = count+1
```

```python
count = 0
while count%2==0:
  print(count)
  count = count+1
```

```python
# loop control statements
##1.break

for i in range(10):
  if i == 5:
    break #it helps to exit the loop
  print(i)
```

```python
#2.continue

for i in range(10):
  if i==5:
    continue#it will skip the particular iteration
  print(i)
```

```python
#3.pass
for i in range(5):
  if i==3:
    print("the number is 3")
    pass
  print(i)
```

```python
#nested loos
for i in range(3):
  for j in range(2):
    print(f"i:{i},j:{j}")
```

```python
#calculate the sum of first n natural numbers

n = int(input("enter the value of n:"))
sum = 0
for i in range(1,n+1):
  sum = sum+i
print("sum of first",n, "natural numbers is",sum)
```

```python
#prime number 1 to100

for num in range(1,101):
  if num>1:
    for i in range(2,num):
      if num%i == 0:
        break
    else:
      print(num)
```
