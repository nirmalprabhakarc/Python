```python
#modeling a bank account
class bank:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance
    def deposit(self,amount):
        self.balance+=amount
    def withdraw(self,amount):
        if amount>self.balance:
            print("not enough balance")
        else:
            self.balance-=amount
    def get_balance(self):
        return self.balance
bank_account = bank('nirmal',5000)
bank_account.get_balance()
```

```python
bank_account.deposit(1000)
bank_account.get_balance()
```

```python
#inheritance
class car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print('the person will drive the',self.enginetype,"car")
        
```

```python
car1 = car(4,5,"petrol")
car1.drive()
```

```python
class tesla(car):
    def __init__(self,windows,doors,enginetype,selfdriving):
        super().__init__(windows,doors,enginetype)
        self.selfdriving1=selfdriving
    def selfdriving(self):
        print("tesla supports",self.selfdriving1)
```

```python
tesla1 = tesla(4,5,"electric","yes")
tesla1.selfdriving()
```

```python
class animal:
    def __init__(self,name):
        self.name = name
    def speak(self):
        print("subclass must implement this")

class pet:
    def __init__(self,owner):
        self.owner = owner

class dog(animal,pet):
    def __init__(self,name,owner):
        animal.__init__(self,name)
        pet.__init__(self,owner)

    def speak(self):
        print(self.name,"is say woof")
```

```python
dog1 = dog('buddy','nirmal')
dog1.speak()
```

```python
##polymorphism with function and methods
class shape:
    def area(self):
        return "the area of figure"

class rectangle(shape):
    def __init__ (self,width,height):
        self.width = width
        self.height = height
    def area(self):
        return self.width*self.height

class circle(shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return self.radius*3.14*self.radius
def print_area(shape):
    print("the area is",shape.area())

retangle1 = rectangle(4,5)
print_area(retangle1)
```

```python
#abstract base class

from abc import ABC,abstractmethod
```

```python
class vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

class car(vehicle):
    def start_engine(self):
        return "car engine started"

class motorcycle(vehicle):
    def start_engine(self):
        return "motor cycle started"
def start_engine(vehicle):
    print(vehicle.start_engine())

car1 = car()
start_engine(car1)
```
