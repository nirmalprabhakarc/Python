```python
#modifying dictionary elements
student = {"name":"nirmal","age":22,"grade":24}
student['age'] = 23
student['address']="india"
print(student)
```

```python
#delete the grade key
del student['grade']
print(student)
```

```python
#common dictionary methods
student = {"name":"nirmal","age":22,"grade":24}
print(student.keys())
print(student.values())
print(student.items())
```

```python
#shallow copy
student = {"name":"nirmal","age":22,"grade":24}
student_copy = student.copy()
print(student_copy)
```

```python
#iterating over dictionarys
student = {"name":"nirmal","age":22,"grade":24}
for key in student.items():
    print(key)
```

```python
#nested dictionary
student = {
    "student1":{
        "name":"nirmal",
        "age":22
    },
    "student2":{
        "name":"nirmal",
        "age":22
    }
}
print(student)
```

```python
#access nested dictionarys
student['student1']['name']
```

```python
#iterating over nested dictionary
for key,value in student.items():
    print(key,value)
    for key1,value1 in value.items():
        print(key1,value1)
```

```python
#dictionary comprehension
square = {num:num*num for num in range(1,11)}
print(square)
```

```python
#conditioning
square = {num:num*num for num in range(1,11) if num%2==0}
print(square)
```

```python
##Practical examples

number = [1,2,2,3,3,3,4,4,4,4]
frequency = {}
for n in number:
  if n in frequency:
    frequency[n] += 1
  else:
    frequency[n] = 1
print(frequency)
```

```python
##merge two dictionarys

dict1 = {"a":1,"b":2}
dict2= {"b":3,"c":4}
merged_dic = {**dict1,**dict2}
print(merged_dic)
```
