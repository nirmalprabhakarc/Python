```python
try:
    num = int(input("enter the number"))
    result = 10/num
except ZeroDivisionError as ex:
    print(ex)
except ValueError as li:
    print(li)
```

```python
#try except and else
try:
    num = int(input("enter the number"))
    result = 10/num
except ZeroDivisionError as ex:
    print(ex)
except ValueError as li:
    print(li)
else:
    print(result)
finally:
    print("executed succesfully")
```

```python
#file and expectional handling
try:
    with open('example1.txt','r') as file:
        content= file.read()
        print(content)
except Exception as ex:
    print(ex)
finally:
    if 'file' in locals() or not file.closed():
        file.close()
        print('file close')
```
