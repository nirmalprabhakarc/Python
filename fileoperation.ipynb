```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
#open a file for writing a file
with open('exapmle.txt','w') as file:
    file.write("hello world\n")
```

```python
#write a file without overwriting
with open('example.txt','a') as file:
    file.write("append")
```

```python
##writing list of lines to a file
lines=['nirmal\n','sai\n','prani\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
#binary files 

with open('example.bin','rb') as file:
    content = file.read()
    print(content)
```

```python
#read the content from a file and write it in another file
with open('example.txt','r') as file:
    content = file.read()
with open('destination.txt','w') as file:
    file.write(content)
```
