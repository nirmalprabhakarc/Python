```python
def even_odd(num):
  '''this is even or odd function'''
  if num % 2 == 0:
    print("The number is even")
  else:
    print("The number is odd")
num =int(input("Enter a number: "))
even_odd(num)#call the function
```

```python
#function with multiple parameters
def add(num1,num2):
  '''This function adds two numbers'''
  sum = num1 + num2
  return sum
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
sum = add(num1,num2)
print("The sum is",sum)
```

```python
##variable length arguments
##positional and keyword argument

def print_num(*nirmal):
  for num in nirmal:
    print(num)
print_num(1,2,3,4,5)
```

```python
#keyword argument

def print_num(**nirmal):
  for key,value in nirmal.items():
    print(key,value)
print_num(a=1,b=2,c=3,d=4,e=5)
```

```python
def con_temp(temp,unit):
  if unit=="celsius":
    return temp* 9/5 +32
  elif unit=="f":
    return (temp-32)*5/9
  else:
    return "none"
print(con_temp(25,'celsius'))
con_temp(77,'f')
```

```python
#password check
def is_str(password):
  if len(password) < 8:
    return False
  if not any(char.isdigit() for char in password):
    return False
  if not any(char.isupper() for char in password):
    return False
  if not any(char.islower() for char in password):
    return False
  return True
password = input("Enter a password: ")
if is_str(password):
  print("Valid password")
else:
  print("Invalid password")
```

```python
#calculating the total cost of items
def cal_total_cost(cart):
  total_cost = 0
  for item in cart:
    total_cost += item["price"]
  return total_cost
cart = [
    {'name':"apple", 'price':1.00},
    {'name':"banana",'price':0.50},
    {'name':"orange",'price':0.75}
]
total_cost = cal_total_cost(cart)
print(total_cost)
```
