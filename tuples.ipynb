```python
#TUPLE METHODS
my_tuple = ('a', 'p', 'p', 'l', 'e',)
print(my_tuple.count('p'))
print(my_tuple.index('l'))
```

```python
#unpacking the tuple
my_tuple = ('John', 'Doe', 25)
name, surname, age = my_tuple
print(name)
print(surname)
print(age)
```

```python
##unpacking with *
my_tuple = (1, 2, 3, 4, 5)
a, b, *c = my_tuple
print(a)
print(b)
print(c)
```

```python
nested_tuple = ((1, 2, 3), (4, 5, 6), (7, 8, 9))
print(nested_tuple[1])
```

```python
#iterating over nested tuple
for sub_tuple in nested_tuple:
    for element in sub_tuple:
        print(element)
```
